## 💰 Part 1: The Hidden Costs of Poor Forecasting

Most businesses don't realize how much money they're losing to inaccurate demand forecasting.

### The Three Core Problems:

1. **Overstock** → Capital tied up in inventory that sits on shelves
2. **Stockouts** → Lost sales when customers can't find what they need
3. **Rush Orders** → Premium pricing to fulfill unexpected demand

Let's calculate what this costs YOUR business:

```python
# Interactive ROI Calculator
def calculate_forecast_roi(
    annual_revenue=2_000_000,
    avg_inventory_days=45,
    current_stockout_rate=0.12,  # 12%
    rush_order_frequency=0.08,   # 8% of orders
    inventory_carrying_cost=0.25,  # 25% annual carrying cost
    rush_order_premium=0.30,     # 30% premium for rush orders
    avg_profit_margin=0.35       # 35% margin
):
    """
    Calculate the financial impact of poor vs. good demand forecasting
    
    Returns dictionary with cost breakdowns and potential savings
    """
    
    # Calculate current costs
    avg_inventory_value = (annual_revenue / 365) * avg_inventory_days
    
    # Cost 1: Excess inventory carrying costs
    # Assume 30% of inventory is overstock due to poor forecasting
    excess_inventory = avg_inventory_value * 0.30
    carrying_cost = excess_inventory * inventory_carrying_cost
    
    # Cost 2: Lost sales from stockouts
    potential_sales = annual_revenue / (1 - current_stockout_rate)
    lost_revenue = potential_sales - annual_revenue
    lost_profit = lost_revenue * avg_profit_margin
    
    # Cost 3: Rush order premiums
    rush_order_cost = annual_revenue * rush_order_frequency * rush_order_premium
    
    total_annual_cost = carrying_cost + lost_profit + rush_order_cost
    
    # Calculate improved scenario (with accurate forecasting)
    # Assumptions:
    # - Reduce excess inventory by 60%
    # - Reduce stockouts by 80%
    # - Reduce rush orders by 70%
    
    improved_carrying_cost = carrying_cost * 0.40
    improved_lost_profit = lost_profit * 0.20
    improved_rush_cost = rush_order_cost * 0.30
    
    improved_total = improved_carrying_cost + improved_lost_profit + improved_rush_cost
    
    annual_savings = total_annual_cost - improved_total
    
    # Assume implementation cost: $15K-25K for small-medium business
    implementation_cost = 20_000
    
    roi = ((annual_savings - implementation_cost) / implementation_cost) * 100
    payback_months = (implementation_cost / annual_savings) * 12
    
    return {
        'current_costs': {
            'excess_inventory_carrying': carrying_cost,
            'lost_profit_stockouts': lost_profit,
            'rush_order_premiums': rush_order_cost,
            'total_annual_cost': total_annual_cost
        },
        'improved_costs': {
            'carrying_cost': improved_carrying_cost,
            'lost_profit': improved_lost_profit,
            'rush_cost': improved_rush_cost,
            'total': improved_total
        },
        'savings': {
            'annual_savings': annual_savings,
            'implementation_cost': implementation_cost,
            'roi_percent': roi,
            'payback_months': payback_months,
            'five_year_value': (annual_savings * 5) - implementation_cost
        }
    }

# Example calculation for a $2M revenue business
results = calculate_forecast_roi(
    annual_revenue=2_000_000,
    current_stockout_rate=0.12,
    avg_inventory_days=45
)

print("="*60)
print("CURRENT ANNUAL COSTS (Poor Forecasting)")
print("="*60)
print(f"Excess Inventory Carrying:  ${results['current_costs']['excess_inventory_carrying']:>12,.0f}")
print(f"Lost Profit (Stockouts):    ${results['current_costs']['lost_profit_stockouts']:>12,.0f}")
print(f"Rush Order Premiums:        ${results['current_costs']['rush_order_premiums']:>12,.0f}")
print(f"{'─'*60}")
print(f"TOTAL ANNUAL COST:          ${results['current_costs']['total_annual_cost']:>12,.0f}")

print("\n" + "="*60)
print("POTENTIAL SAVINGS (Accurate Forecasting)")
print("="*60)
print(f"Annual Savings:             ${results['savings']['annual_savings']:>12,.0f}")
print(f"Implementation Cost:        ${results['savings']['implementation_cost']:>12,.0f}")
print(f"First Year ROI:             {results['savings']['roi_percent']:>13,.0f}%")
print(f"Payback Period:             {results['savings']['payback_months']:>12,.1f} months")
print(f"5-Year Value:               ${results['savings']['five_year_value']:>12,.0f}")
```

### 📈 Visualization: Cost Breakdown

```python
# Visualize cost comparison
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Current costs breakdown
current_costs = results['current_costs']
cost_categories = ['Excess\nInventory', 'Lost\nSales', 'Rush\nOrders']
cost_values = [
    current_costs['excess_inventory_carrying'],
    current_costs['lost_profit_stockouts'],
    current_costs['rush_order_premiums']
]

colors = ['#e74c3c', '#e67e22', '#f39c12']
ax1.bar(cost_categories, cost_values, color=colors, alpha=0.8)
ax1.set_ylabel('Annual Cost ($)', fontsize=12, fontweight='bold')
ax1.set_title('Current Cost of Poor Forecasting', fontsize=14, fontweight='bold')
ax1.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'${x/1000:.0f}K'))

# Add value labels on bars
for i, (cat, val) in enumerate(zip(cost_categories, cost_values)):
    ax1.text(i, val, f'${val:,.0f}', ha='center', va='bottom', fontweight='bold')

# Before vs After comparison
comparison_data = {
    'Current\n(Poor Forecast)': results['current_costs']['total_annual_cost'],
    'Improved\n(Accurate Forecast)': results['improved_costs']['total']
}

bars = ax2.bar(comparison_data.keys(), comparison_data.values(), 
               color=['#e74c3c', '#27ae60'], alpha=0.8, width=0.6)
ax2.set_ylabel('Total Annual Cost ($)', fontsize=12, fontweight='bold')
ax2.set_title('Before vs. After: Total Impact', fontsize=14, fontweight='bold')
ax2.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'${x/1000:.0f}K'))

# Add value labels and savings annotation
for bar, val in zip(bars, comparison_data.values()):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
             f'${val:,.0f}', ha='center', va='bottom', fontweight='bold', fontsize=11)

# Add savings annotation
savings = results['savings']['annual_savings']
ax2.annotate('', xy=(1, results['improved_costs']['total']), 
             xytext=(1, results['current_costs']['total_annual_cost']),
             arrowprops=dict(arrowstyle='<->', color='#2ecc71', lw=2))
ax2.text(1.15, (results['current_costs']['total_annual_cost'] + results['improved_costs']['total'])/2,
         f'Save\n${savings:,.0f}', fontweight='bold', fontsize=12, color='#27ae60',
         bbox=dict(boxstyle='round', facecolor='white', edgecolor='#27ae60', linewidth=2))

plt.tight_layout()
plt.savefig('../../visualizations/cost_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

print(f"\n💡 Bottom Line: You could save ${savings:,.0f} per year")
print(f"   That's {(savings/results['current_costs']['total_annual_cost'])*100:.0f}% reduction in forecast-related costs!")
```

## 🏪 Part 3: Real Saskatchewan Business Example

### Case Study: Regina Retail Chain (8 Locations)

Let's look at real data from a Saskatchewan retail business and see the impact of forecasting.

```python
# Load the sample Saskatchewan retail data
df = pd.read_csv('../../data/sample_retail_sales.csv', parse_dates=['date'])

print("📊 Saskatchewan Retail Data Overview")
print("="*60)
print(f"Date Range: {df['date'].min().date()} to {df['date'].max().date()}")
print(f"Total Days: {len(df)}")
print(f"\nSales Statistics:")
print(f"  Average Daily Sales: ${df['sales'].mean():,.2f}")
print(f"  Minimum Daily Sales: ${df['sales'].min():,.2f}")
print(f"  Maximum Daily Sales: ${df['sales'].max():,.2f}")
print(f"  Standard Deviation:  ${df['sales'].std():,.2f}")
print(f"\nTotal Revenue (2 years): ${df['sales'].sum():,.2f}")

# Show sample of the data
print("\n📋 Sample Data:")
display(df.head(10))
```

```python
# Visualize the demand patterns
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Sales over time
ax = axes[0, 0]
ax.plot(df['date'], df['sales'], linewidth=1, alpha=0.7, color='#3498db')
ax.set_title('Daily Sales Over Time', fontsize=12, fontweight='bold')
ax.set_xlabel('Date')
ax.set_ylabel('Sales ($)')
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'${x/1000:.0f}K'))
ax.grid(True, alpha=0.3)

# Highlight key patterns
ax.axvspan(pd.Timestamp('2022-12-01'), pd.Timestamp('2022-12-31'), 
           alpha=0.2, color='green', label='Holiday Season')
ax.axvspan(pd.Timestamp('2023-12-01'), pd.Timestamp('2023-12-31'), 
           alpha=0.2, color='green')
ax.legend()

# 2. Seasonal patterns
ax = axes[0, 1]
seasonal_avg = df.groupby('season')['sales'].mean().sort_values()
colors_season = {'winter': '#3498db', 'spring': '#2ecc71', 'summer': '#f39c12', 'fall': '#e67e22'}
bars = ax.barh(seasonal_avg.index, seasonal_avg.values, 
               color=[colors_season[s] for s in seasonal_avg.index])
ax.set_title('Average Sales by Season', fontsize=12, fontweight='bold')
ax.set_xlabel('Average Daily Sales ($)')
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'${x/1000:.0f}K'))

# Add value labels
for i, (season, value) in enumerate(seasonal_avg.items()):
    ax.text(value, i, f' ${value:,.0f}', va='center', fontweight='bold')

# 3. Day of week patterns
ax = axes[1, 0]
dow_names = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
dow_avg = df.groupby('day_of_week')['sales'].mean()
bars = ax.bar(dow_names, dow_avg.values, color='#9b59b6', alpha=0.8)
ax.set_title('Average Sales by Day of Week', fontsize=12, fontweight='bold')
ax.set_xlabel('Day of Week')
ax.set_ylabel('Average Sales ($)')
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'${x/1000:.0f}K'))

# Highlight weekend
bars[5].set_color('#e74c3c')  # Saturday
bars[6].set_color('#e67e22')  # Sunday

for i, val in enumerate(dow_avg.values):
    ax.text(i, val, f'${val:,.0f}', ha='center', va='bottom', fontsize=9, fontweight='bold')

# 4. Temperature impact
ax = axes[1, 1]
temp_bins = pd.cut(df['temperature'], bins=[-50, -20, -10, 0, 10, 20, 40])
temp_sales = df.groupby(temp_bins)['sales'].mean()
temp_labels = ['< -20°C\n(Extreme)', '-20 to -10°C\n(Very Cold)', 
               '-10 to 0°C\n(Cold)', '0 to 10°C\n(Cool)', 
               '10 to 20°C\n(Mild)', '> 20°C\n(Warm)']

bars = ax.bar(range(len(temp_sales)), temp_sales.values, color='#16a085', alpha=0.8)
ax.set_xticks(range(len(temp_sales)))
ax.set_xticklabels(temp_labels, fontsize=9)
ax.set_title('Sales by Temperature (Saskatchewan Climate)', fontsize=12, fontweight='bold')
ax.set_ylabel('Average Sales ($)')
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'${x/1000:.0f}K'))

# Highlight extreme cold impact
if len(bars) > 0:
    bars[0].set_color('#e74c3c')

for i, val in enumerate(temp_sales.values):
    ax.text(i, val, f'${val:,.0f}', ha='center', va='bottom', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.savefig('../../visualizations/demand_patterns.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n📍 Key Saskatchewan-Specific Insights:")
print("  • Extreme cold (-20°C+) reduces sales significantly")
print("  • Strong December holiday peak")
print("  • Weekend shopping dominance (especially Saturday)")
print("  • Spring has highest average sales (post-winter spending)")
```
